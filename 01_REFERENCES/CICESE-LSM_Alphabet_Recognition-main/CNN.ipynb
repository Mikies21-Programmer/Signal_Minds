La primera arquitectura será la base para partir al respecto, será inicialmente una **Matriz** de 2x21 que tendrá dos inputs para las CNN, será en las columnas los keypoints (1,2,3,...,21) y en los renglones la ubicación en el eje X o Y

Algo con esta estructura:

| Keypoint | 1   | 2   | 3   | ... | 21  |
|----------|-----|-----|-----|-----|-----|
| X coord  | x1  | x2  | x3  | ... | x21 |
| Y coord  | y1  | y2  | y3  | ... | y21 |

Todo esto será para un solo frame, pero conocemos perfectamente que no todos los modelos están manejando a un solo frame, entonces será esta estructura n cantidad de veces por frames en el video.

Entonces quedará como un tensor de **Frames x Coordenadas x Keypoints**

Esto permitirá tener una extensión temporal, que será adecuada para CNNs

Y la segunda, con una arquitectura de 21 capas, una por cada keypoint que existe dentro de los frames.
